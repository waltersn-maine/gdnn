```python
# paths
SLOPE = "../data/constructs/sledG_torch_pitedge_pitpit_2010.npz"
BASIN = "../data/constructs/basin_mushgraph2010.npz"
LINKS = "../data/constructs/pit_basin_links2010.npz"
```

```python
import numpy as np, CGTN_torchdata as D
from sklearn.ensemble import HistGradientBoostingRegressor
A = D.load_artifacts(SLOPE, BASIN, LINKS); out = {}
for s in [0, 13, 42]:
    tr, va, te = D.make_split(A, s)
    g = HistGradientBoostingRegressor(max_iter=400, learning_rate=0.06, l2_regularization=1.0, random_state=s)
    g.fit(A["X_raw"][tr.numpy()], A["y"].numpy()[tr.numpy()])
    out[f"pred_s{s}"] = g.predict(A["X_raw"]).astype(np.float32)   # all nodes
    out[f"test_s{s}"] = te.numpy()
np.savez("../results/gbt_preds.npz", y=A["y"].numpy(), node_rc=A["node_rc"], **out)
```
