```python
import geopandas as gpd, pandas as pd
s = gpd.read_file("../../data/outlines/ncanwell_longslope_nolap.gpkg")
b = gpd.read_file("../../data/outlines/ncanwell_longbasin_nolap.gpkg")
geom = pd.concat([s, b], ignore_index=True).union_all()   # use .unary_union on geopandas < 0.14
dom = gpd.GeoDataFrame(geometry=[geom], crs=s.crs)
dom.to_file("../../data/outlines/ncanwell_fulldomain.gpkg", driver="GPKG")
print("domain CRS:", dom.crs, "| area km²:", round(dom.area.iloc[0] / 1e6, 2))
```
