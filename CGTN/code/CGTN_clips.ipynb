```python
import rioxarray as rxr, geopandas as gpd
from rasterio.enums import Resampling

dom = gpd.read_file("../data/outlines/ncanwell_fulldomain.gpkg")

# reference grid = 2010 DEM clipped to the full domain
ref = rxr.open_rasterio("../data/zarchive2010/cfc_elevfilled_2010_5m.tif", masked=True).rio.clip(dom.geometry, dom.crs)
ref.rio.to_raster("../data/rasters/canwell_domainlong_elev2010_5m.tif")

layers = [
  ("cfc_slope_2010.tif",      "slope2010_5m",      Resampling.bilinear),
  ("cfc_aspect_2010.tif",     "aspect2010_5m",     Resampling.nearest),    # nearest: no averaging across the 0/360 wrap
  ("cfc_roughness_2010.tif",  "roughness2010_5m",  Resampling.bilinear),
  ("cfc_tri_2010.tif",        "tri2010_5m",        Resampling.bilinear),
  ("cfc_tpi_2010.tif",        "tpi2010_5m",        Resampling.bilinear),
  ("dist_basin_2010.tif", "dist_basin2010_5m", Resampling.bilinear),
  ("ndvi_filled_lsat05_2008_2012_0601_0901.tif", "ndvi2010_5m", Resampling.bilinear),   # 30 m -> 5 m
  ("lst_filled_lsat05_2008_2012_0601_0901.tif",  "lst2010_5m",  Resampling.bilinear),   # 30 m -> 5 m
  ("cfc_ddem_2010_2000_coreg_5m.tif",               "ddem_2010_2000_5m",   Resampling.bilinear),  # feature (confirm/build)
  ("canwell_ddem_2025_2010_coreg_5m.tif", "ddem_2025_2010_5m", Resampling.bilinear),  # TARGET
]
for path, name, rs in layers:
    r = (rxr.open_rasterio("../data/zarchive2010/" + path, masked=True)
            .rio.reproject_match(ref, resampling=rs)
            .rio.clip(dom.geometry, dom.crs))
    r.rio.to_raster("../data/rasters/canwell_domainlong_%s.tif" % name)
    print(f"{name:12s} {tuple(r.shape[-2:])}  valid {int(r.notnull().sum()):,}")
```
