```python
# quick notebook to prepare data for Claude's processing

# 1. load + feature matrix
import numpy as np, pandas as pd

df = pd.read_parquet("../data/constructs1/slope_feature_table.parquet")
FEATS = ["elev","slope","aspect_sin","aspect_cos","roughness",
         "tri","tpi","ndvi","lst","dist"]
X = df[FEATS].to_numpy(np.float64)
print("rows:", X.shape[0], "| features:", X.shape[1],
      "| NaNs:", int(np.isnan(X).sum()))
```

```python
import numpy as np
FEATS=["elev","slope","aspect_sin","aspect_cos","roughness","tri","tpi","ndvi","lst","dist"]
np.savez_compressed("../data/constructs1/slope_table.npz",
    X=df[FEATS].to_numpy("float32"), y=df["ddem"].to_numpy("float32"),
    slope_id=df["slope_id"].to_numpy("int16"),
    row=df["row"].to_numpy("int32"), col=df["col"].to_numpy("int32"))

print("npz saved")
```
