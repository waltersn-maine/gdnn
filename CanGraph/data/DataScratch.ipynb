```python
import geopandas as gpd
from shapely.geometry import Polygon

def strip_z(geom):
    return Polygon([(x, y) for x, y, z in geom.exterior.coords])

north = gpd.read_file('CanwellNorthSlopeFull_utm.gpkg')
basin = gpd.read_file('Canwell2009Basin_utm.gpkg')

north['geometry'] = north['geometry'].apply(strip_z)
basin['geometry'] = basin['geometry'].apply(strip_z)

print("North has_z:", north.geometry.has_z.any())
print("Basin has_z:", basin.geometry.has_z.any())

north.to_file('CanwellNorthSlope_clean_utm.gpkg', driver='GPKG', layer='domain')
basin.to_file('CanwellBasin_clean_utm.gpkg', driver='GPKG', layer='domain')
```
